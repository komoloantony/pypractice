```python
import pandas as pd
music_data = pd.read_csv("music_dataset_.csv")
music_data
```

```python
print(music_data.dtypes)
music_data.head()
music_data["Average_Listener_Age"] = pd.to_numeric(music_data["Average_Listener_Age"], errors="coerce")
music_data["Duration_Minutes"] = pd.to_numeric(music_data["Duration_Minutes"], errors="coerce")
music_data["Year"] = pd.to_numeric(music_data["Year"], errors="coerce")
music_data.dropna()
```

```python
X = music_data.drop(columns={"Song","Artist","Genre","Popularity","Typical_Age_Group"})#Input cell
X
```

```python
y = music_data[["Song","Artist","Genre","Popularity","Typical_Age_Group"]]#output cell
y
```

```python
from sklearn.tree import DecisionTreeClassifier

X = music_data[["Year","Duration_Minutes","Average_Listener_Age"]]
y = music_data["Song"]
model = DecisionTreeClassifier()
model.fit(X, y)
predictions = model.predict([[2020,3.8,20],
                             [2017,2.1,19]])

print(predictions)
```
