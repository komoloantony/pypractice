```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

student = pd.read_csv('students.csv')
X = student.drop(columns=["Result"])
y = student["Result"]

model = DecisionTreeClassifier()
model.fit(X, y)

name = input("Enter name of the student: ").title()
hours_studied = input("Enter the number of hours studied: ")
attendace = input("Enter students attendance: ")
assingments = input("Enter number of assignments done: ")
hours_slept = input("Enter the number of hours slept: ")

prediction = model.predict([[hours_studied, attendace, assingments, hours_slept]])
print(name,prediction)
```

# Lets do an accuracy calculation

```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier

student = pd.read_csv('students.csv')
X = student.drop(columns=["Result"])
y = student["Result"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.8)

model = DecisionTreeClassifier()
model.fit(X_train, y_train)
model.fit(X_test, y_test)
predictions = model.predict(X_test)

score = accuracy_score(y_test, predictions)
score
```
